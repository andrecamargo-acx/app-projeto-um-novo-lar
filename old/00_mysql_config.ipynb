
# ⚙️ 00 - Configuração do MySQL

Este notebook é usado para **cadastrar os parâmetros de conexão** com o banco de dados MySQL.

Ele grava um arquivo `mysql_config.json` na mesma pasta, que será utilizado
pelos outros notebooks para abrir a conexão.

> **Importante:** nunca versionar esse arquivo com senha real em repositório público.


```python

import json
from pathlib import Path

CONFIG_PATH = Path("mysql_config.json")

def salvar_config_mysql(
    host: str,
    port: int,
    user: str,
    password: str,
    database: str
):
    """Salva as configurações de conexão em um arquivo JSON local."""
    cfg = {
        "host": host,
        "port": port,
        "user": user,
        "password": password,
        "database": database
    }
    CONFIG_PATH.write_text(json.dumps(cfg, indent=2), encoding="utf-8")
    print(f"[OK] Configuração salva em: {CONFIG_PATH.resolve()}")

def carregar_config_mysql() -> dict:
    """Carrega as configurações de conexão do arquivo JSON."""
    if not CONFIG_PATH.exists():
        raise FileNotFoundError(
            f"Arquivo de configuração {CONFIG_PATH} não encontrado. "
            "Execute a função salvar_config_mysql() primeiro."
        )
    return json.loads(CONFIG_PATH.read_text(encoding="utf-8"))

# Configuração já preenchida para o seu ambiente MySQL
salvar_config_mysql(
    host="bd-unl.mysql.uhserver.com",
    port=3306,
    user="usr_unl",
    password="Usrunl0159.",
    database="bd_unl"
)
```
